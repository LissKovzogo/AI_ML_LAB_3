```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 11, 1)

y1 = np.array([(-0.2)*i**2 + 2*i for i in x])
y2 = np.array([(-0.4)*i**2 + 4*i for i in x])
y3 = np.array([2*i for i in x])

labels = ["y1", "y2", "y3"]


fig, ax = plt.subplots()
ax.stackplot(x, y1, y2, y3, labels=labels)
ax.set_xlabel(r'$x$')     
ax.set_ylabel(r'$y$')
ax.grid(True)
ax.legend(loc='upper left')

plt.show()
```

```python
import matplotlib.colors as mcolors
bc = mcolors.BASE_COLORS
x = np.arange(0, 10.5, 0.5)
y=np.cos(x)
num_set=np.random.randint(1, len(mcolors.BASE_COLORS), len(x))
sizes=num_set*35
colors = [list(bc.keys())[i] for i in num_set]
plt.scatter(x,y, s=sizes,c=colors,linewidths=2, edgecolors='face')
plt.plot(x,y,'g--',alpha=0.4)
plt.xlabel(r'$x$')
plt.ylabel(r'$y$')
plt.grid(True)
plt.show()
```

```python
vals = [24, 17, 53, 21, 35]
labels = [r"\textbf{Ford}", r"\textbf{Toyota}", r"\textbf{BMW}", r"\textbf{AUDI}", r"\textbf{Jaguar}"]

fig, ax = plt.subplots(figsize=(8, 8))
ax.pie(vals, labels=labels, autopct=r'%1.1f\%%', textprops={'fontsize': 12})
ax.axis("equal")
ax.set_title(r'\textbf{Car Sales Distribution}', fontsize=14)

plt.show()
```

```python
vals = [24, 17, 53, 21, 35]

labels = ["Ford", "Toyota", "BMW", "AUDI", "Jaguar"]
explode = (0.1, 0, 0.15, 0, 0) 

fig, ax = plt.subplots()

ax.pie(vals, labels=labels, autopct='%1.1f%%', shadow=True, explode=explode, wedgeprops={'lw':1, 'ls':'--', 'edgecolor':"k"}, rotatelabels=True)

ax.axis("equal")
plt.show()
```

```python
fig, ax = plt.subplots()

offset = 0.4

data = np.array([[5, 10, 7], [8, 15, 5], [11, 9, 7]])
cmap = plt.get_cmap("tab20b")

b_colors = cmap(np.array([0, 8, 12]))
sm_colors = cmap(np.array([1, 2, 3, 9, 10, 11, 13, 14, 15]))

ax.pie(data.sum(axis=1), radius=1, colors=b_colors, wedgeprops=dict(width=offset, edgecolor='w'))
ax.pie(data.flatten(), radius=1-offset, colors=sm_colors, wedgeprops=dict(width=offset, edgecolor='w'))

ax.set_title(r'$Pie\ Chart$', fontsize=14)
plt.show()
```

```python

np.random.seed(19680801)
data = np.random.randn(25, 25)

plt.imshow(data, cmap='viridis')  
plt.xlabel(r'$x$ --- \textbf{pixel index}', fontsize=12)
plt.ylabel(r'$y$ --- \textbf{pixel index}', fontsize=12)
plt.title(r'\textbf{Random 2D Array Heatmap}', fontsize=14)
plt.colorbar(label=r'\textbf{Value}')
plt.grid(True, linestyle='--', alpha=0.3)  

plt.show()
```

```python

fig, axs = plt.subplots(1, 2, figsize=(10, 3), constrained_layout=True)
p1 = axs[0].imshow(data, cmap='winter', aspect='equal', vmin=-1, vmax=1, origin='lower')
axs[0].set_xlabel(r'$x$', fontsize=12)
axs[0].set_ylabel(r'$y$', fontsize=12)
fig.colorbar(p1, ax=axs[0])

p2 = axs[1].imshow(data, cmap='plasma', aspect='equal', interpolation='gaussian', origin='lower', extent=(0, 30, 0, 30))
axs[1].set_xlabel(r'$x$', fontsize=12)
axs[1].set_ylabel(r'$y$', fontsize=12)
fig.colorbar(p2, ax=axs[1])

plt.show()
```

```python
np.random.seed(123)
data = np.random.rand(5,7)
plt.pcolormesh(data, cmap='plasma', edgecolors='k', shading='flat')

plt.xlabel(r'$x$')
plt.ylabel(r'$y$')
plt.title(r'\textbf{Pcolormesh plot}')
plt.grid(True)
plt.colorbar(label=r'\textbf{Value}')

plt.show()
```

```python
x = np.linspace(-np.pi, np.pi,50)
y=x
z = np.cos(x)
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.set_xlabel(r'$x$', fontsize=12)
ax.set_ylabel(r'$y$', fontsize=12)
ax.set_zlabel(r'$z = \cos(x)$', fontsize=12)
ax.grid(True, linestyle='--', alpha=0.5)

ax.plot(x,y,z, label='parametric curve')
plt.show()
```
